## Project Phase 04 — Data Transformation and Management
### Difficulty: Intermediate


```python
import numpy as np
import pandas as pd
d=pd.read_excel("Online_Retail_Practice.xlsx")
d=pd.DataFrame(d)
df_raw=d.copy()
df=df_raw.copy()
df
```

```python
df.rename(columns={"Description":"ProductDescription"},inplace=True)
df
```

```python
df["Revenue"]=df["Quantity"]*df["UnitPrice"]
df
```

```python
df["TransactionType"] = np.select(
    [
        df["InvoiceNo"].astype("string").str.startswith("C", na=False),
        df["Quantity"] < 0,
        df["Quantity"] > 0,
        (df["Quantity"] == 0) | df["Quantity"].isna()
    ],
    [
        "cancellation",
        "return",
        "positive sale",
        "other/invalid"
    ],
    default="other/invalid"
)
df
```

```python
df["IsIdentifiedCustomer"] = df["CustomerID"].notnull()
df
```

```python
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"], errors="coerce")
df
```

```python
for col in ["InvoiceNo", "StockCode", "Country"]:
    df[col] = df[col].astype("string")
df
```

```python
first_cols = [
    "InvoiceNo",
    "InvoiceDate",
    "StockCode",
    "ProductDescription",
    "CustomerID",
    "Country",
    "Quantity",
    "UnitPrice",
    "Revenue",
    "TransactionType",
    "IsIdentifiedCustomer"
]
ordered_cols=first_cols+[
    col for col in df.columns if col not in first_cols
]
df=df[ordered_cols]
df
```

```python
df = df.sort_values(
    by=["InvoiceDate", "InvoiceNo"],
    na_position="last"
)
df
```

```python
df.reset_index(drop=True, inplace=True)
df
```

```python
print(d[d["CustomerID"].isnull()])
print(d["CustomerID"].isnull().sum())
```

```python
print(d[d["Description"].isnull() |
        d["Description"].str.strip().eq("")])
```

```python
report = pd.DataFrame({
    "Missing": d.isnull().sum(),
    "NonMissing": d.notnull().sum(),
    "Percentage": d.isnull().mean() * 100
})
print(report)
```

```python
df = d.drop_duplicates().copy()
print("Before:", len(d))
print("After:", len(df))
```
